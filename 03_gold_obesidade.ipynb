```python
%sql

CREATE OR REPLACE TABLE gold_fastfood_obesidade AS

SELECT
    consumo_fastfood,
    COUNT(*) AS total_pessoas,
    SUM(
        CASE
            WHEN classe_obesidade = 'Obesidade' THEN 1
            ELSE 0
        END
    ) AS qtd_obesos,
    ROUND(
        100.0 * SUM(
            CASE
                WHEN classe_obesidade = 'Obesidade' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS percentual_obesidade

FROM silver_obesidade_tratada

GROUP BY consumo_fastfood

ORDER BY consumo_fastfood;
```

```python
%sql
SELECT *
FROM gold_fastfood_obesidade;
```

```python
%sql

CREATE OR REPLACE TABLE gold_perfil_obesidade AS

SELECT
    sexo,
    faixa_etaria,
    nivel_atividade AS atividade_fisica,
    consumo_fastfood,
    frequencia_vegetais AS consumo_vegetais,
    historico_familiar,
    nivel_uso_tecnologia AS tempo_tecnologia,
    consumo_liquidos AS consumo_agua,
    classe_obesidade,
    COUNT(*) AS quantidade

FROM silver_obesidade_tratada

WHERE classe_obesidade = 'Obesidade'

GROUP BY
    sexo,
    faixa_etaria,
    nivel_atividade,
    consumo_fastfood,
    frequencia_vegetais,
    historico_familiar,
    nivel_uso_tecnologia,
    consumo_liquidos,
    classe_obesidade

ORDER BY quantidade DESC;
```
