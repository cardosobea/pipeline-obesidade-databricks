```python
%sql

CREATE OR REPLACE TABLE silver_obesidade_tratada AS

SELECT
    CASE
        WHEN Sex = 1 THEN 'Masculino'
        WHEN Sex = 2 THEN 'Feminino'
    END AS sexo,

    Age AS idade,

    Height AS altura_cm,

    CASE
        WHEN Overweight_Obese_Family = 1 THEN 'Sim'
        WHEN Overweight_Obese_Family = 2 THEN 'Não'
    END AS historico_familiar,

    CASE
        WHEN Consumption_of_Fast_Food = 1 THEN 'Sim'
        WHEN Consumption_of_Fast_Food = 2 THEN 'Não'
    END AS consumo_fastfood,

    CASE
        WHEN Frequency_of_Consuming_Vegetables = 1 THEN 'Raramente'
        WHEN Frequency_of_Consuming_Vegetables = 2 THEN 'Às vezes'
        WHEN Frequency_of_Consuming_Vegetables = 3 THEN 'Sempre'
    END AS frequencia_vegetais,

    CASE
        WHEN Number_of_Main_Meals_Daily = 1 THEN '1-2'
        WHEN Number_of_Main_Meals_Daily = 2 THEN '3'
        WHEN Number_of_Main_Meals_Daily = 3 THEN '3+'
    END AS refeicoes_principais_dia,

    CASE
        WHEN Food_Intake_Between_Meals = 1 THEN 'Raramente'
        WHEN Food_Intake_Between_Meals = 2 THEN 'Às vezes'
        WHEN Food_Intake_Between_Meals = 3 THEN 'Geralmente'
        WHEN Food_Intake_Between_Meals = 4 THEN 'Sempre'
    END AS alimentacao_entre_refeicoes,

    CASE
        WHEN Smoking = 1 THEN 'Sim'
        WHEN Smoking = 2 THEN 'Não'
    END AS fumante,

    CASE
        WHEN Liquid_Intake_Daily = 1 THEN 'Menos de 1 litro'
        WHEN Liquid_Intake_Daily = 2 THEN 'Entre 1 e 2 litros'
        WHEN Liquid_Intake_Daily = 3 THEN 'Mais de 2 litros'
    END AS consumo_liquidos,

    CASE
        WHEN Calculation_of_Calorie_Intake = 1 THEN 'Sim'
        WHEN Calculation_of_Calorie_Intake = 2 THEN 'Não'
    END AS controle_calorias,

    CASE
        WHEN Physical_Excercise = 1 THEN 'Sem atividade'
        WHEN Physical_Excercise = 2 THEN '1-2 dias'
        WHEN Physical_Excercise = 3 THEN '3-4 dias'
        WHEN Physical_Excercise = 4 THEN '5-6 dias'
        WHEN Physical_Excercise = 5 THEN '6+ dias'
    END AS atividade_fisica,

    CASE
        WHEN Schedule_Dedicated_to_Technology = 1 THEN '0-2 horas'
        WHEN Schedule_Dedicated_to_Technology = 2 THEN '3-5 horas'
        WHEN Schedule_Dedicated_to_Technology = 3 THEN 'Mais de 5 horas'
    END AS uso_tecnologia,

    CASE
        WHEN Type_of_Transportation_Used = 1 THEN 'Automóvel'
        WHEN Type_of_Transportation_Used = 2 THEN 'Motocicleta'
        WHEN Type_of_Transportation_Used = 3 THEN 'Bicicleta'
        WHEN Type_of_Transportation_Used = 4 THEN 'Transporte público'
        WHEN Type_of_Transportation_Used = 5 THEN 'Caminhada'
    END AS meio_transporte,

    CASE
        WHEN Class = 1 THEN 'Abaixo do peso'
        WHEN Class = 2 THEN 'Normal'
        WHEN Class = 3 THEN 'Sobrepeso'
        WHEN Class = 4 THEN 'Obesidade'
    END AS classe_obesidade,

    CASE
        WHEN Age BETWEEN 18 AND 25 THEN '18-25'
        WHEN Age BETWEEN 26 AND 35 THEN '26-35'
        WHEN Age BETWEEN 36 AND 45 THEN '36-45'
        WHEN Age BETWEEN 46 AND 54 THEN '46-54'
        ELSE 'Fora do intervalo'
    END AS faixa_etaria,

    CASE
        WHEN Physical_Excercise = 1 THEN 'Sedentário'
        WHEN Physical_Excercise IN (2, 3) THEN 'Moderado'
        WHEN Physical_Excercise IN (4, 5) THEN 'Ativo'
        ELSE 'Não classificado'
    END AS nivel_atividade,

    CASE
        WHEN Schedule_Dedicated_to_Technology = 1 THEN 'Baixo'
        WHEN Schedule_Dedicated_to_Technology = 2 THEN 'Médio'
        WHEN Schedule_Dedicated_to_Technology = 3 THEN 'Alto'
        ELSE 'Não classificado'
    END AS nivel_uso_tecnologia

FROM bronze_obesidade_raw;
```

```python
%sql
SELECT COUNT(*) AS total_registros
FROM silver_obesidade_tratada;
```

```python
%sql
SELECT *
FROM silver_obesidade_tratada
LIMIT 10;
```

```python
%sql
SELECT
    COUNT(*) AS total_registros,
    COUNT(*) - COUNT(sexo) AS nulos_sexo,
    COUNT(*) - COUNT(idade) AS nulos_idade,
    COUNT(*) - COUNT(altura_cm) AS nulos_altura,
    COUNT(*) - COUNT(historico_familiar) AS nulos_historico_familiar,
    COUNT(*) - COUNT(consumo_fastfood) AS nulos_fastfood,
    COUNT(*) - COUNT(frequencia_vegetais) AS nulos_vegetais,
    COUNT(*) - COUNT(atividade_fisica) AS nulos_atividade,
    COUNT(*) - COUNT(consumo_liquidos) AS nulos_liquidos,
    COUNT(*) - COUNT(meio_transporte) AS nulos_transporte,
    COUNT(*) - COUNT(classe_obesidade) AS nulos_classe
FROM silver_obesidade_tratada;
```
